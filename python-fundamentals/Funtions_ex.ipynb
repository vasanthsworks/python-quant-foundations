Write a function called introduce_person() that accepts three parameters: first_name, last_name, and age.
The function should print a sentence such as: “My name is John Doe and I am 25 years old.”
Call the function once using positional arguments.
Call the function a second time using keyword arguments.
Finally, call the function again with the positional arguments in the wrong order and observe the output.

```python
def introduce_person(fn, ln, age):
    print("My name is", fn, ln, "and I am", age, "years old.")
age = int(input("enter your age: "))
fn = input("enter your first name: ")
ln = input("enter your last name: ")
introduce_person(fn, ln, age)
```

8-8. User Albums: Start with your program from Exercise 8-7. Write a while
loop that allows users to enter an album’s artist and title. Once you have that
information, call make_album() with the user’s input and print the dictionary
that’s created. Be sure to include a quit value in the while loop.

```python
#8.7
def make_album(artist, title, tracks=None):
    album = {'artist': artist, 'title': title}
    if tracks:
        album['tracks'] = tracks
    return album
album1 = make_album("Post Malone", "Sunflower")
album2 = make_album("Hans Zimmer", "Cornfield Chase", tracks=1)
album3 = make_album("Juice WRLD", "Wasted", tracks=16)
print(album1)
print(album2)
print(album3)

#8.8
def make_album(artist, title, tracks=None):
    album = {'artist': artist, 'title': title}
    if tracks:
        album['tracks'] = tracks
    return album
while True:
    print("\nPlease enter album details (or 'q' to quit at any time):")
    artist_name = input("Artist name: ")
    if artist_name.lower() == 'q':
        break
    title_name = input("Album title: ")
    if title_name.lower() == 'q':
        break     
    user_album = make_album(artist_name, title_name)
    print(user_album)
```
