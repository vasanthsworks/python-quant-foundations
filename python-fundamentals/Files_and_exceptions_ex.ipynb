Clean & Parse (.strip()): Given a raw list of incoming ticker signals:

Python
raw_signals = ["  NVDA \n", " MSFT  \n", "AAPL\n", "  AMZN "]

Clean each ticker using .strip() and store them cleanly.
Audit Logging (Append Mode 'a'):

Write a function log_executions(filename, tickers) that takes a target filename and your cleaned tickers.
Open the file in append mode ('a') so prior trading days aren't overwritten.
Format each entry with a timestamp/execution tag, e.g.:
[EXEC] NVDA
[EXEC] MSFT
Log Reader & Line Parsing (readlines()):
Write a function read_audit_log(filename) that:
Opens the file in read mode ('r').
Uses .readlines() to capture every log line into a Python list.
Iterates through the list and prints only lines containing "NVDA"

```python
raw_signals = ["  NVDA \n", " MSFT  \n", "AAPL\n", "  AMZN "]
clean_signals=[s.strip() for s in raw_signals]
def log_executions(filename,tickers):
    with open(filename,'a') as file:
        for ticker in tickers:
            file.write(f"[EXEC] {ticker}\n")
log_executions('executions_logger.txt',clean_signals)
def read_audit_log(filename):
    try:
        with open(filename,'r') as file:
            lines=file.readlines()
    except FileNotFoundError:
        print(f"[WARN] Log file {filename} not found. Creating a new entry...")
    else:
        for line in lines:
                if "NVDA" in line:
                    print(line.rstrip())
    finally:
        print("[SYS] Audit attempt finished")
read_audit_log('executions_logger.txt')
print(clean_signals)

```
