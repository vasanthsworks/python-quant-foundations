The Challenge — Market Data Processor

You are building the first component of a small quantitative trading system.

A market-data feed provides the following information for six consecutive timestamps:

Mid-prices: 245.20, 246.10, 244.80, 247.35, 248.00, 247.60

Bid prices: 245.10, 246.00, 244.70, 247.25, 247.90, 247.50

Ask prices: 245.30, 246.20, 244.90, 247.45, 248.10, 247.70

Volume: 1200, 850, 2100, 1750, 950, 1400

Your strategy holds 200 shares.

Using NumPy:

Create a NumPy array for each of the four market-data series.
Calculate the bid-ask spread at every timestamp.
Calculate the value of your 200-share position at every timestamp using the mid-price.
Calculate the notional traded at every timestamp using:
Notional=Price×Volume

```python
#q1 Create a NumPy array for each of the four market-data series.
import numpy as np
mid_prices = np.array([245.20, 246.10, 244.80, 247.35, 248.00, 247.60])
bid_prices = np.array([245.10, 246.00, 244.70, 247.25, 247.90, 247.50])
ask_prices = np.array([245.30, 246.20, 244.90, 247.45, 248.10, 247.70])
volume = np.array([1200, 850, 2100, 1750, 950, 1400])
#Q2 Calculate the bid-ask spread at every timestamp.
spread =  ask_prices - bid_prices 
#Q3 Calculate the value of your 200-share position at every timestamp using the mid-price.
value = mid_prices * 200 
#Q4 Calculate the notional traded at every timestamp
notional_value = mid_prices *volume
print(spread)
print(value)
print(notional_value)
```
