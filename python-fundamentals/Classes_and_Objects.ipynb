Your Requirements:
Define the __init__ constructor that accepts three parameters:

ticker (string, e.g. "AAPL")

shares (int, e.g. 10)

avg_price (float, e.g. 150.0)

Assign all three to self attributes!

Write a method called get_current_value(self, current_price):

It takes current_price as a parameter.

It calculates and returns the total market value of that position (shares * current_price).

Write a method called get_pnl(self, current_price):

P&L stands for Profit and Loss.

It calculates and returns total profit or loss: (current_price - avg_price) * shares.

```python
class Position:
    def __init__ (self,ticker,shares,avg_price):
       self.shares=shares
       self.ticker=ticker
       self.avg_price=avg_price
    def get_current_value(self,current_price):
       total_value=self.shares *current_price
       return total_value
    def get_pnl(self,current_price):
        total_money_made=(current_price-self.avg_price)*self.shares
        print("total money made in this trade is",total_money_made)
        return total_money_made
aapl_trade=Position("AAPL",10,150.0)
value=aapl_trade.get_current_value(150)
profit=aapl_trade.get_pnl(175)
print(f"Current Value: ${value}")  
print(f"Profit/Loss:   ${profit}")   

```
